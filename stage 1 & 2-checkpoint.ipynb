```python
# ============================================================
# SPE AFRICA GEOTHERMAL DATATHON 2026
# TASK 1 & 2: SUBSURFACE ASSESSMENT + SURFACE SYSTEM DESIGN
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import os
import glob
import warnings
warnings.filterwarnings("ignore")

# Data handling
import numpy as np
import pandas as pd

# Visualization
import matplotlib.pyplot as plt

# For Excel files
import openpyxl

try:
    import lasio
    print("lasio imported successfully.")
except ImportError:
    print("lasio is not installed. Run: pip install lasio")
```

```python
# -----------------------------
# 2. DEFINE PROJECT PATHS
# -----------------------------

# Current notebook working directory
PROJECT_PATH = os.getcwd()

print("Current working directory:")
print(PROJECT_PATH)

# Main data folder
BASE_PATH = os.path.join(PROJECT_PATH, "data_AGD_2026")

# Data subfolder
DATA_PATH = os.path.join(BASE_PATH, "data")

# Raw LAS files
RAW_PATH = os.path.join(DATA_PATH, "raw")

print("\nBase path:", BASE_PATH)
print("Data path:", DATA_PATH)
print("Raw LAS path:", RAW_PATH)

# -------------------------------------------------
# CHECK FILES INSIDE DATA DIRECTORY
# -------------------------------------------------

print("\nFiles inside data folder:")
for file in os.listdir(DATA_PATH):
    print("-", file)

# -------------------------------------------------
# CHECK LAS FILES
# -------------------------------------------------

las_files = glob.glob(os.path.join(RAW_PATH, "*.las"))

print("\nNumber of LAS files found:", len(las_files))

for file in las_files:
    print("-", os.path.basename(file))
```

```python
# ---------------------------------------------------------
# 3. LOAD EXCEL / CSV DATASETS
# ---------------------------------------------------------

# File paths
litho_file = os.path.join(DATA_PATH, "Lithostratigraphic Data.xlsx")
thermogis_file = os.path.join(DATA_PATH, "ThermoGIS Data.xlsx")
wellpath_file = os.path.join(DATA_PATH, "Well Path Data.xlsx")
target_lithologies_file = os.path.join(DATA_PATH, "target_lithologies.csv")

# Load datasets
litho_df = pd.read_excel(litho_file)
thermogis_df = pd.read_excel(thermogis_file)
wellpath_df = pd.read_excel(wellpath_file)
target_lithologies_df = pd.read_csv(target_lithologies_file)

print("Datasets loaded successfully.")
```

```python
# ---------------------------------------------------------
# 4. QUICK DATA PREVIEW
# ---------------------------------------------------------

print("=" * 80)
print("LITHOSTRATIGRAPHIC DATA")
display(litho_df.head())

print("=" * 80)
print("THERMOGIS DATA")
display(thermogis_df.head())

print("=" * 80)
print("WELL PATH DATA")
display(wellpath_df.head())

print("=" * 80)
print("TARGET LITHOLOGIES")
display(target_lithologies_df.head())
```

```python
# ---------------------------------------------------------
# 5. DATASET INFORMATION
# ---------------------------------------------------------

datasets = {
    "Lithostratigraphic": litho_df,
    "ThermoGIS": thermogis_df,
    "Well Path": wellpath_df,
    "Target Lithologies": target_lithologies_df
}

for name, df in datasets.items():
    
    print("\n" + "=" * 80)
    print(name)
    
    print("\nShape:")
    print(df.shape)
    
    print("\nColumns:")
    print(df.columns.tolist())
    
    print("\nMissing Values:")
    display(df.isna().sum())
```

```python
# ---------------------------------------------------------
# 6. LOAD LAS WELL LOG FILES
# ---------------------------------------------------------

las_data = {}

for file in las_files:
    
    well_name = os.path.basename(file).replace(".las", "")
    
    try:
        # Read LAS file
        las = lasio.read(file)
        
        # Convert to dataframe
        df = las.df().reset_index()
        
        # Rename depth column
        df = df.rename(columns={df.columns[0]: "DEPTH"})
        
        # Add well name
        df["WELL"] = well_name
        
        # Store dataframe
        las_data[well_name] = df
        
        print(f"{well_name} loaded successfully")
        print("Shape:", df.shape)
        print("Columns:", df.columns.tolist())
        print("-" * 80)
        
    except Exception as e:
        print(f"Error loading {well_name}: {e}")
```

```python
# ---------------------------------------------------------
# 7. COMBINE WELL LOG DATA
# ---------------------------------------------------------

well_log_df = pd.concat(las_data.values(), ignore_index=True)

print("Combined well log shape:", well_log_df.shape)

print("\nAvailable wells:")
print(well_log_df["WELL"].unique())

print("\nColumns in combined log dataset:")
print(well_log_df.columns.tolist())

print("\nMissing values:")
display(well_log_df.isna().sum())

display(well_log_df.head())
```

```python
# ---------------------------------------------------------
# 8. CHECK AVAILABLE LOG CURVES IN EACH WELL
# ---------------------------------------------------------

for well_name, df in las_data.items():
    
    print("\n" + "=" * 80)
    print("WELL:", well_name)
    
    print("\nColumns / Curves:")
    print(df.columns.tolist())
    
    print("\nDepth range:")
    print(df["DEPTH"].min(), "to", df["DEPTH"].max())
    
    print("\nShape:")
    print(df.shape)
```

```python
# ---------------------------------------------------------
# 9. FIND COMMON LOG CURVES
# ---------------------------------------------------------

all_curves = []

for well_name, df in las_data.items():
    curves = set(df.columns.tolist())
    all_curves.append(curves)

common_curves = set.intersection(*all_curves)

print("Common curves across all wells:")
print(sorted(common_curves))
```

```python
# ---------------------------------------------------------
# 10. PLOT GAMMA RAY LOGS
# ---------------------------------------------------------

def plot_log(df, curve, title):
    
    plt.figure(figsize=(4,8))
    
    plt.plot(df[curve], df["DEPTH"], color="green")
    
    plt.gca().invert_yaxis()
    
    plt.xlabel(curve)
    plt.ylabel("Depth (m)")
    plt.title(title)
    
    plt.grid(True)
    plt.show()

# Plot GR for each well if available
for well_name, df in las_data.items():
    
    if "GR" in df.columns:
        
        plot_log(
            df,
            "GR",
            f"{well_name} - Gamma Ray"
        )
```

```python
# ---------------------------------------------------------
# 11. STANDARDIZE WELL LOG CURVES
# ---------------------------------------------------------

standardized_logs = {}

for well_name, df in las_data.items():
    
    temp = df.copy()
    
    # -----------------------------------------
    # Standardize sonic log names
    # -----------------------------------------
    
    if "DTC" in temp.columns:
        temp["DT"] = temp["DTC"]
    
    # -----------------------------------------
    # Remove duplicate depth columns
    # -----------------------------------------
    
    if "DEPT:2" in temp.columns:
        temp = temp.drop(columns=["DEPT:2"])
    
    # -----------------------------------------
    # Keep useful geothermal curves only
    # -----------------------------------------
    
    useful_cols = [
        col for col in [
            "DEPTH",
            "GR",
            "RHOB",
            "DT",
            "NPHI",
            "DRHO",
            "WELL"
        ]
        if col in temp.columns
    ]
    
    temp = temp[useful_cols]
    
    standardized_logs[well_name] = temp
    
    print(f"{well_name}")
    print(temp.columns.tolist())
    print(temp.shape)
    print("-" * 60)
```

```python
# ---------------------------------------------------------
# 12. COMBINE STANDARDIZED LOGS
# ---------------------------------------------------------

logs_df = pd.concat(
    standardized_logs.values(),
    ignore_index=True
)

print("Combined standardized logs shape:")
print(logs_df.shape)

print("\nColumns:")
print(logs_df.columns.tolist())

print("\nMissing values:")
display(logs_df.isna().sum())

display(logs_df.head())
```

```python
# ---------------------------------------------------------
# 13. CLEAN WELL LOG DATA
# ---------------------------------------------------------

# Remove rows where all major logs are missing
logs_df = logs_df.dropna(
    subset=["GR", "RHOB", "DT"],
    how="all"
)

# Reset index
logs_df = logs_df.reset_index(drop=True)

print("Cleaned logs shape:")
print(logs_df.shape)

print("\nMissing values after cleaning:")
display(logs_df.isna().sum())

display(logs_df.head())
```

```python
# ---------------------------------------------------------
# 14. INSPECT TARGET LITHOLOGY DATA
# ---------------------------------------------------------

print("Columns:")
print(target_lithologies_df.columns.tolist())

print("\nUnique wells:")
print(target_lithologies_df["well_id"].unique())

print("\nFormation thickness statistics:")
display(
    target_lithologies_df["formation_thickness_m"].describe()
)

print("\nFormation top depth statistics:")
display(
    target_lithologies_df["formation_top_tvd"].describe()
)

print("\nFormation base depth statistics:")
display(
    target_lithologies_df["formation_base_tvd"].describe()
)
```

```python
# ---------------------------------------------------------
# 15. EXTRACT TARGET RESERVOIR INTERVALS
# ---------------------------------------------------------

reservoir_logs = []

for well in target_lithologies_df["well_id"].unique():
    
    # Get well logs
    temp_logs = logs_df[
        logs_df["WELL"] == well
    ].copy()
    
    # Skip if no logs
    if len(temp_logs) == 0:
        continue
    
    # Get formation interval
    temp_target = target_lithologies_df[
        target_lithologies_df["well_id"] == well
    ]
    
    top = temp_target["formation_top_tvd"].iloc[0]
    base = temp_target["formation_base_tvd"].iloc[0]
    
    # Extract reservoir interval
    interval = temp_logs[
        (temp_logs["DEPTH"] >= top) &
        (temp_logs["DEPTH"] <= base)
    ].copy()
    
    interval["TOP"] = top
    interval["BASE"] = base
    
    reservoir_logs.append(interval)
    
    print(f"{well}")
    print(f"Top: {top}")
    print(f"Base: {base}")
    print(f"Thickness: {base-top}")
    print(f"Extracted rows: {len(interval)}")
    print("-" * 60)
```

```python
# ---------------------------------------------------------
# 16. COMBINE RESERVOIR INTERVALS
# ---------------------------------------------------------

reservoir_df = pd.concat(
    reservoir_logs,
    ignore_index=True
)

print("Reservoir interval dataset shape:")
print(reservoir_df.shape)

display(reservoir_df.head())
```

```python
# ---------------------------------------------------------
# 17. RESERVOIR INTERVAL SUMMARY PER WELL
# ---------------------------------------------------------

reservoir_summary = (
    reservoir_df
    .groupby("WELL")
    .agg(
        top_depth_m=("TOP", "first"),
        base_depth_m=("BASE", "first"),
        thickness_m=("DEPTH", lambda x: x.max() - x.min()),
        avg_GR_api=("GR", "mean"),
        min_GR_api=("GR", "min"),
        max_GR_api=("GR", "max"),
        avg_RHOB_gcc=("RHOB", "mean"),
        avg_DT_usft=("DT", "mean"),
        avg_NPHI_frac=("NPHI", "mean"),
        data_points=("DEPTH", "count")
    )
    .reset_index()
)

display(reservoir_summary)
```

```python
# ---------------------------------------------------------
# 18. NET SAND AND NET-TO-GROSS ESTIMATION
# ---------------------------------------------------------

GR_CUTOFF = 75

reservoir_df["is_net_sand"] = (reservoir_df["GR"] <= GR_CUTOFF).astype(int)

net_sand_summary = (
    reservoir_df
    .groupby("WELL")
    .agg(
        total_points=("is_net_sand", "count"),
        net_sand_points=("is_net_sand", "sum"),
        net_to_gross=("is_net_sand", "mean"),
        avg_GR_api=("GR", "mean")
    )
    .reset_index()
)

# Approximate net thickness
thickness_map = reservoir_summary.set_index("WELL")["thickness_m"].to_dict()
net_sand_summary["gross_thickness_m"] = net_sand_summary["WELL"].map(thickness_map)
net_sand_summary["net_sand_thickness_m"] = (
    net_sand_summary["net_to_gross"] * net_sand_summary["gross_thickness_m"]
)

display(net_sand_summary)
```

```python
# ---------------------------------------------------------
# 19. PLOT GR AND RHOB IN RESERVOIR INTERVAL
# ---------------------------------------------------------

def plot_reservoir_logs(well):
    temp = reservoir_df[reservoir_df["WELL"] == well].copy()
    
    fig, axes = plt.subplots(1, 3, figsize=(12, 8), sharey=True)
    
    axes[0].plot(temp["GR"], temp["DEPTH"])
    axes[0].axvline(GR_CUTOFF, linestyle="--")
    axes[0].set_xlabel("GR (API)")
    axes[0].set_ylabel("Depth (m)")
    axes[0].set_title("Gamma Ray")
    axes[0].grid(True)
    
    axes[1].plot(temp["RHOB"], temp["DEPTH"])
    axes[1].set_xlabel("RHOB (g/cc)")
    axes[1].set_title("Bulk Density")
    axes[1].grid(True)
    
    axes[2].plot(temp["DT"], temp["DEPTH"])
    axes[2].set_xlabel("DT")
    axes[2].set_title("Sonic")
    axes[2].grid(True)
    
    axes[0].invert_yaxis()
    fig.suptitle(f"{well} - Rotliegend Reservoir Interval", fontsize=14)
    plt.show()

for well in reservoir_df["WELL"].unique():
    plot_reservoir_logs(well)
```

```python
# ---------------------------------------------------------
# 20. ESTIMATE DENSITY POROSITY
# ---------------------------------------------------------

# Matrix density for sandstone
RHO_MATRIX = 2.65

# Fluid density (water)
RHO_FLUID = 1.0

# Density porosity equation
reservoir_df["PHI_D"] = (
    (RHO_MATRIX - reservoir_df["RHOB"]) /
    (RHO_MATRIX - RHO_FLUID)
)

# Limit realistic values
reservoir_df["PHI_D"] = reservoir_df["PHI_D"].clip(0, 0.4)

print("Density porosity calculated.")

display(
    reservoir_df[
        ["WELL", "DEPTH", "RHOB", "PHI_D"]
    ].head()
)
```

```python
# ---------------------------------------------------------
# 21. POROSITY SUMMARY
# ---------------------------------------------------------

porosity_summary = (
    reservoir_df
    .groupby("WELL")
    .agg(
        avg_density_porosity=("PHI_D", "mean"),
        min_density_porosity=("PHI_D", "min"),
        max_density_porosity=("PHI_D", "max"),
        avg_nphi=("NPHI", "mean")
    )
    .reset_index()
)

display(porosity_summary)
```

```python
# ---------------------------------------------------------
# 22. CREATE WELL-LEVEL RESERVOIR METRICS
# ---------------------------------------------------------

well_metrics = reservoir_summary.merge(
    porosity_summary,
    on="WELL",
    how="left"
)

display(well_metrics)
```

```python
# ---------------------------------------------------------
# 23. NORMALIZE RESERVOIR PARAMETERS
# ---------------------------------------------------------

# Lower GR is better
well_metrics["GR_score"] = (
    1 - (
        (well_metrics["avg_GR_api"] - well_metrics["avg_GR_api"].min()) /
        (well_metrics["avg_GR_api"].max() - well_metrics["avg_GR_api"].min())
    )
)

# Higher porosity is better
well_metrics["PHI_score"] = (
    (well_metrics["avg_density_porosity"] - well_metrics["avg_density_porosity"].min()) /
    (well_metrics["avg_density_porosity"].max() - well_metrics["avg_density_porosity"].min())
)

# Higher thickness is better
well_metrics["THICKNESS_score"] = (
    (well_metrics["thickness_m"] - well_metrics["thickness_m"].min()) /
    (well_metrics["thickness_m"].max() - well_metrics["thickness_m"].min())
)

display(
    well_metrics[
        [
            "WELL",
            "GR_score",
            "PHI_score",
            "THICKNESS_score"
        ]
    ]
)
```

```python
# ---------------------------------------------------------
# 24. RESERVOIR QUALITY INDEX (RQI)
# ---------------------------------------------------------

# Weighted reservoir quality score
well_metrics["RQI"] = (
    0.4 * well_metrics["PHI_score"] +
    0.3 * well_metrics["GR_score"] +
    0.3 * well_metrics["THICKNESS_score"]
)

# Rank wells
well_metrics = well_metrics.sort_values(
    "RQI",
    ascending=False
)

display(
    well_metrics[
        [
            "WELL",
            "avg_GR_api",
            "avg_density_porosity",
            "thickness_m",
            "RQI"
        ]
    ]
)
```

```python
# ---------------------------------------------------------
# 25. PLOT RESERVOIR QUALITY INDEX
# ---------------------------------------------------------

plt.figure(figsize=(8,5))

plt.bar(
    well_metrics["WELL"],
    well_metrics["RQI"]
)

plt.ylabel("Reservoir Quality Index")
plt.title("Geothermal Reservoir Quality Ranking")

plt.grid(axis="y")

plt.show()
```

```python
# ---------------------------------------------------------
# 26. GEOTHERMAL SYSTEM PARAMETERS
# ---------------------------------------------------------

# Flow rate
flowrate_lps = 50

# Convert to kg/s
# Assuming water density ≈ 1000 kg/m3
mass_flowrate = flowrate_lps

# Specific heat capacity of water
cp_water = 4250  # J/kg.K

# Reservoir production temperature
T_reservoir = 94  # °C

# Reinjection / surface temperature
T_reinject = 10  # °C

# Temperature difference
delta_T = T_reservoir - T_reinject

print("Mass flowrate:", mass_flowrate, "kg/s")
print("Delta T:", delta_T, "°C")
```

```python
# ---------------------------------------------------------
# 27. THERMAL POWER CALCULATION
# ---------------------------------------------------------

# Thermal power in Watts
P_th_W = (
    mass_flowrate *
    cp_water *
    delta_T
)

# Convert to MWth
P_th_MWth = P_th_W / 1e6

print("Estimated Thermal Power:")
print(round(P_th_MWth, 2), "MWth")
```

```python
# ---------------------------------------------------------
# 28. COOLING CAPACITY CHECK
# ---------------------------------------------------------

cooling_requirement_MWth = 5
heating_requirement_MWth = 10

heating_surplus = P_th_MWth - heating_requirement_MWth

print("Available geothermal heat:", round(P_th_MWth, 2), "MWth")
print("Heating requirement:", heating_requirement_MWth, "MWth")
print("Heating surplus:", round(heating_surplus, 2), "MWth")

if P_th_MWth >= heating_requirement_MWth:
    print("Heating requirement satisfied.")
else:
    print("Heating requirement not satisfied.")

# Assume cooling is provided through an absorption/electric chiller assisted by geothermal heat
# Simple conservative assumption: 5 MWth cooling is feasible if geothermal heat exceeds the heating base requirement
if heating_surplus >= cooling_requirement_MWth:
    print("Cooling requirement can be supported from geothermal surplus.")
else:
    print("Cooling may require heat pump/chiller support or thermal storage.")
```

```python
# ---------------------------------------------------------
# 29. DEFINE SURFACE SYSTEM CONFIGURATION
# ---------------------------------------------------------

surface_system = {
    
    "production_well": 1,
    
    "injection_well": 1,
    
    "heat_exchanger": True,
    
    "heat_pump": True,
    
    "district_heating_network": True,
    
    "district_cooling_network": True,
    
    "thermal_storage": True,
    
    "backup_system": "auxiliary boiler",
    
    "cooling_system": "absorption/electric chiller"
}

print("Surface geothermal system configuration:")
print()

for key, value in surface_system.items():
    print(f"{key}: {value}")
```

```python
# ---------------------------------------------------------
# 30. DISTRICT ENERGY DISTRIBUTION
# ---------------------------------------------------------

# District demand
heating_demand = 10      # MWth
cooling_demand = 5       # MWth

# Available geothermal heat
available_heat = P_th_MWth

# Remaining energy after heating
remaining_heat = available_heat - heating_demand

print("Available geothermal heat:", round(available_heat, 2), "MWth")
print("Heating demand:", heating_demand, "MWth")
print("Remaining heat:", round(remaining_heat, 2), "MWth")

# Cooling support
if remaining_heat >= cooling_demand:
    
    cooling_supported = cooling_demand
    
else:
    
    cooling_supported = remaining_heat

print("Cooling supported:", round(cooling_supported, 2), "MWth")
```

```python
# ---------------------------------------------------------
# 31. HEAT PUMP PERFORMANCE
# ---------------------------------------------------------

# Coefficient of performance
COP = 4.0

# Electrical energy needed
heat_pump_power = heating_demand / COP

print("Heat pump COP:", COP)

print(
    "Estimated electrical power required:",
    round(heat_pump_power, 2),
    "MWe"
)
```

```python
# ---------------------------------------------------------
# 32. THERMAL STORAGE ESTIMATION
# ---------------------------------------------------------

# Assume 6-hour storage reserve
storage_hours = 6

# Thermal energy storage needed
thermal_storage_MWh = (
    heating_demand *
    storage_hours
)

print(
    "Estimated thermal storage requirement:",
    thermal_storage_MWh,
    "MWh"
)
```

```python
# ---------------------------------------------------------
# 33. DISTRICT SYSTEM EFFICIENCY
# ---------------------------------------------------------

# Assume network efficiency
network_efficiency = 0.90

usable_heat = (
    available_heat *
    network_efficiency
)

print("District network efficiency:", network_efficiency)

print(
    "Usable district heat:",
    round(usable_heat, 2),
    "MWth"
)
```

```python
# ---------------------------------------------------------
# 34. SURFACE SYSTEM SUMMARY
# ---------------------------------------------------------

surface_summary = pd.DataFrame({
    
    "Parameter": [
        "Available geothermal heat",
        "Heating demand",
        "Cooling demand",
        "Heating surplus",
        "Heat pump COP",
        "Heat pump power requirement",
        "Thermal storage capacity",
        "District efficiency",
        "Usable district heat"
    ],
    
    "Value": [
        round(available_heat, 2),
        heating_demand,
        cooling_demand,
        round(remaining_heat, 2),
        COP,
        round(heat_pump_power, 2),
        thermal_storage_MWh,
        network_efficiency,
        round(usable_heat, 2)
    ],
    
    "Unit": [
        "MWth",
        "MWth",
        "MWth",
        "MWth",
        "-",
        "MWe",
        "MWh",
        "-",
        "MWth"
    ]
})

display(surface_summary)
```

```python
# ---------------------------------------------------------
# 35. SURFACE SYSTEM FEASIBILITY CHECK
# ---------------------------------------------------------

total_demand = heating_demand + cooling_demand

print("Total heating + cooling demand:", total_demand, "MWth")
print("Usable geothermal heat:", round(usable_heat, 2), "MWth")

if usable_heat >= total_demand:
    print("System feasibility: PASS")
    print("The proposed geothermal surface system can meet both heating and cooling demand.")
else:
    print("System feasibility: FAIL")
    print("Additional backup or hybrid energy support is required.")

capacity_margin = usable_heat - total_demand

print("Capacity margin:", round(capacity_margin, 2), "MWth")
```
